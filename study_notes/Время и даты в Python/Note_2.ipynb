Пример ошибки внутри функции

```python
def square_sum(*args):
    total_sum = 0
    for arg in args:
        total_sum += arg**2

    return total_sum
```

```python
square_sum(1, 2, 3)
```

```python
# пытаемся применить операцию возведения в квадрат к строке
# ----> 5 square_sum(1, 2, '3') - в какой функции произошла ошибка
# ----> 4     total_sum += arg**2 - в какой именно строке произошла ошибка

square_sum(1, 2, '3')
```

## Как сделать, чтобы цикл с расчетом не падал каждый раз

```python
try:
    # ваш код, где может произойти ошибка
    float('123a')

except:
    # код который выполняется в случае ошибки
    print('Кривая строка')
    pass
```

```python
data = ['90', '60', '90', '240tot']
total_sum = 0

for num in data:
    try:
        total_sum += float(num)

    except:
        print('Ошибка в данных: {}'.format(num))

print('Итого', total_sum)
```

Как сохранить всю информацию об ошибке?

```python
# полная версия traceback
import traceback

try:
    float('123fff')

except Exception:
    print(traceback.print_exc())

print('Проехали!')
```
