- 'r' открыть для чтения. Файл при этом не меняется

## Контекстный менеджер
Как обезопасить себя от подобныхх накладок

```python
with open('results.csv', 'w') as f:
    f.write('Начинаю запись первой строки...\n')
    f.write('Начинаю запись второй строки...\n')

    my_friend_results = open('results.csv', 'w')
    my_friend_results.write('Тут еще результаты есть')
    my_friend_results.close()
```

## А можно читать и сразу в файл записывать?
Напишите функцию, которая фильрует файл visit_log (2).csv по источнуку context и пишет результат в visits_context.csv. Используйте функцию из второго упражнения для проверки результата.

```python
with open('visit_log (2).csv', 'r') as f:
    with open('visits_context.csv', 'w') as f2write:
        for line in f:
            if 'context' in line:
                f2write.write(line)
```

## Модуль pickle
Запись объекта сразу в файл как потом байтов

```python
import pickle
```

```python
data = {'user_id': '1840e0b9d4', 'category': 'Продукты'}
```

```python
with open('data.pickle', 'wb') as f:
    pickle.dump(data, f)
```

```python
# прочитать объект из такого файла

with open('data.pickle', 'rb') as f:
    dict_ = pickle.load(f)

dict_, dict_['user_id']
```
